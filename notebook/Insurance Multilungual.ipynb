```python
import os
import docx
import pdfplumber
from deep_translator import GoogleTranslator
import joblib
```

```python
def extract_text_and_translate_multiple(file_path, src_lang="en", dest_langs=None):
    if dest_langs is None:
        dest_langs = ["ta", "hi", "te", "kn", "ml", "bn", "gu", "mr"]  # 8 Indian languages

    ext = os.path.splitext(file_path)[-1].lower()

    # Extract text based on file type
    if ext == ".txt":
        with open(file_path, "r", encoding="utf-8") as f:
            text = f.read()
    elif ext == ".docx":
        doc = docx.Document(file_path)
        text = "\n".join([para.text for para in doc.paragraphs])
    elif ext == ".pdf":
        text = ""
        with pdfplumber.open(file_path) as pdf:
            for page in pdf.pages:
                page_text = page.extract_text()
                if page_text:
                    text += page_text + "\n"
    else:
        raise ValueError("❌ Unsupported file type!")

    lines = [line.strip() for line in text.split("\n") if line.strip()]

    # Translate into all target languages
    translated_results = {}
    for lang in dest_langs:
        try:
            translator = GoogleTranslator(source=src_lang, target=lang)
            translated_text = [translator.translate(line) for line in lines]
            translated_results[lang] = "\n".join(translated_text)
        except Exception as e:
            translated_results[lang] = f"⚠️ Translation failed for {lang}: {str(e)}"

    return translated_results
```

```python
# dump the model
joblib.dump(extract_text_and_translate_multiple, "insurance_translate.pkl")
```
